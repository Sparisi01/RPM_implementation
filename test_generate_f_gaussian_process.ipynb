```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def GP_generate(n : int ,N_samples : int = 1,low : float = 0, high : float = 1, l : float = 0.1, sigma : float = 1) -> np.ndarray:
    
    # Space array
    x_array = np.linspace(low, high, n)

    # Gaussian kernel
    def kernel(x1, x2, l: np.array=0.1, sigma: np.ndarray = 1):
        sqdist = (x1[:, None] - x2[None, :])**2
        return (sigma*sigma)*np.exp(-sqdist / (2 * l**2))

    # Covariance Matrix
    K = kernel(x_array, x_array, l=l, sigma=sigma)
 
    #K += 1e-8 * np.eye(n)

    # Sample
    u = np.random.multivariate_normal(mean=np.zeros(n), cov=K, size=N_samples)
    
    return u
```
