```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def H(z,h,L):
    return 2*z / ((z + h)*np.exp(-z*L)+(z - h)*np.exp(z*L))

def G(z,h,L):
    return 2*np.sinh(z) / ((z + h)*np.exp(-z*L)+(z - h)*np.exp(z*L))
```

```python
L, h, D = 1, 1, 1

omegas = np.arange(1,1000,1)
z = np.sqrt(omegas/(2*D))*complex(1,-1)

plt.plot(omegas,abs(H(z,h,L)),label=r"|H($\omega$)|")
plt.plot(omegas,abs(G(z,h,L)),label=r"|G($\omega$)|")

plt.xscale("log")
plt.yscale("log")
plt.xlabel(r'$\omega$')
plt.ylabel(r'Transfer function ABS')
plt.legend()
```
